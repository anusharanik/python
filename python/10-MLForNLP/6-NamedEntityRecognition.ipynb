Named Entity Recognition

```python
sentence="The Eiffel Tower, located in Paris, France, is one of the most famous landmarks in the world. It was designed by the engineer Gustave Eiffel and completed in 1889 as the entrance arch for the 1889 World's Fair.Standing at about 330 meters tall, it attracts millions of visitors each year from countries such as the United States, Japan, and India. At night, the tower sparkles with thousands of lights, offering a breathtaking view over the River Seine. UNESCO declared it a World Heritage Site, and it remains a symbol of French culture and architectural brilliance."
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
words
```

```python
tag_elements=nltk.pos_tag(words)
```
