Applications: Q&A, Text summarization, chatbot

```python
import nltk
nltk.download('punkt')
#nltk.download('wordnet')
```

```python
##Wordnet Lemmatizer
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
'''
POS Noun-n
adjective-a
verb-v
adverb-r
'''
lemmatizer.lemmatize('going',pos='v')
```

```python
##Python list
words = [
    "cats", "dogs", "children", "mice", "feet", "leaves", "geese",
    "running", "ate", "gone", "playing", "studies", "written", "taking","history",
    "better", "worse", "fastest", "happier"
]
```

```python
for word in words:
    print(word+"-------->"+lemmatizer.lemmatize(word,pos='n'))
```
