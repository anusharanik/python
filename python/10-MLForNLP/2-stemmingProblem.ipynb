Stemming--PorterStemmer Class

```python
words=["eating","eats","eaten","writing","writes","written","history"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

RegexpStemmer Class

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

Snowball Stemmer

```python
from nltk.stem import SnowballStemmer
```

```python
snowballstemmer=SnowballStemmer('english')
```

```python
for word in words:
    print(word+"----------->"+snowballstemmer.stem(word))
```

```python
#PorterStemmer gives meaningless output for some words
stemming.stem("fairly"),stemming.stem("sportingly")
```

```python
snowballstemmer.stem("fairly"),snowballstemmer.stem("sportingly")
```

```python
''''The stemming process gives meaningless output for some words. 
This is the limitation of Stemming. No chatbaots use this technique.
'''

snowballstemmer.stem("goes")
```
