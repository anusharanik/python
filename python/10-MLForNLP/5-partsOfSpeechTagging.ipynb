```python
paragraph = """
My dear young friends, I have always believed that dreams are not what you see when you sleep; dreams are those that do not let you sleep. 
Each one of us has a purpose in life, and we must work hard to achieve it with perseverance and faith. 
When I was a child, I had a dream to become a pilot, but destiny had other plans for me. 
I became a scientist, and through that path, I was able to serve my nation and contribute to the success of our space and missile programs. 
Great dreams of great dreamers are always transcended. 
If you want to shine like the sun, first burn like the sun. 
Failure will never overtake you if your determination to succeed is strong enough. 
Let us all take an oath to make our nation strong, prosperous, and peaceful. 
We must have courage to think differently, courage to invent, courage to travel the unexplored path, and courage to discover the impossible. 
This is the spirit of youth. When you face problems, remember that difficulties are not to make you weak but to make you stronger. 
Success does not come to you without effort. You must sweat, you must work, and you must sacrifice for your goal. 
Remember, the ignited mind of the youth is the most powerful resource on the earth, above the earth, and under the earth.
"""
```

```python
import nltk
nltk.download('punkt')
#nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
from nltk.corpus import stopwords
sentences=nltk.sent_tokenize(paragraph)
```

```python
sentences
```

```python
##We will find out the POS tag

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[word for word in words if word not in set(stopwords.words('english'))]
    pos_tag=nltk.pos_tag(words)
    print(pos_tag)
```
