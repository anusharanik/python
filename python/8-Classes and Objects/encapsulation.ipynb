Encapsulation

```python
#Encapsulation with Getter and Setter methods
##Acess Modifiers:public, protected and private variables

class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age #Both name and age are public variables
def get_name(person):
    return person.name
person=Person("User",34)
print(person.name)
get_name(person)
```

```python
dir(person)
```

```python
#Private variables
class Person:
    def __init__(self,name,age,gender):
        self.__name=name
        self.__age=age #Both name and age are private variables
        self.gender=gender

def get_name(person):
    return person.__name

person=Person("User",34,"Male")

get_name(person)
```

```python
dir(person)
```

```python
#Protected variables
class Person:
    def __init__(self,name,age,gender):
        self._name=name
        self._age=age #Both name and age are protected variables
        self.gender=gender
class Employee(Person):
    def __init__(self,name,age,gender):
        super().__init__(name,age,gender)

employee=Employee("EmpName",34,"Male")
print(employee._name)
```

```python
##Encapsulation with Getter and Setter methods
class Person:
    def __init__(self,name,age):
        self.__name=name #Private variable
        self.__age=age   #Private variable
    
    ##Getter method for name
    def get_name(self):
        return self.__name
    
    ##Setter method for name
    def set_name(self,name):
        self.__name=name

    #Getter method for age
    def get_age(self):
        return self.__age
    
    #Setter method for age
    def set_age(self,age):
        if age>0:
            self.__age=age
        else:
            print("Age cannot be negative.")

person=Person("UserName",34)

##Access and modify private variable using setter and getter methods

print(person.get_name())
print(person.get_age())

person.set_age(45)
print(person.get_age())

person.set_age(-56)
print(person.get_age())


```
