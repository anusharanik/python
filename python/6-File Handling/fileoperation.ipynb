```python
## Read a whole file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
# Read a file line by line
with open('example.txt','r')as file:
    for line in file:
        print(line.strip()) #strip() removes the newline character
```

```python
## Writing to a file (Overwriting)

with open('example.txt','w')as file:
    file.write("Adding a new line\n")
    file.write("second line added\n")
```

```python
##Writing to  a file (without overwriting)

with open('example.txt','a')as file:
    file.write("Appends without overwriting\n")
```

```python
##Write multiple lines to a file

lines=['First line\n','Second line\n','third line\n']
with open('example.txt','a')as file:
    file.writelines(lines)
```

```python
## Write to a binary file
data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
## Reading a binary file
with open('example.bin','rb')as file:
    content=file.read()
    print(content)
```

```python
## Read the content from source file and write into a destination file
with open('example.txt','r')as source_file:
    content=source_file.read()

with open('destination.txt','w')as destination_file:
    destination_file.write(content)    
```

```python
#Read a text file and count number of words, characters and lines
line_count=0;
word_count=0;
char_count=0;
with open('example.txt',"r") as file:
    for line in file:
        line_count=line_count+1
        words=line.split()
        word_count=word_count+len(words)
        char_count=char_count+len(line)
print(f"Number of lines in a file:{line_count}")
print(f"Number of words in a file:{word_count}")
print(f"Number of characters in a file:{char_count}")
```

```python
##Write into a file and then read from the file
with open('example.txt','w+')as file:
    file.write("This is the first line\n")
    file.write("This is the second line\n")

    # Move the file cursor to the beginning
    file.seek(0)
    content=file.read()
    print(content)

## Read the file
```
